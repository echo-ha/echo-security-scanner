```python
import sqlalchemy
import pandas as pd

engine = sqlalchemy.create_engine("postgresql://scanner:scanner@localhost:5432/scanner")
```

```python
# List all tables
with engine.connect() as conn:
    tables = pd.read_sql("SELECT tablename FROM pg_tables WHERE schemaname = 'public'", conn)
tables
```

```python
with engine.connect() as conn:
    images = pd.read_sql("SELECT * FROM images", conn)
images
```

```python
with engine.connect() as conn:
    cves = pd.read_sql("SELECT * FROM cves", conn)
print(f"{len(cves)} rows")
cves.head(20)
```
